```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])

df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df.head()
```

```python
df.dtypes
```

```python
df["Hospital"] = df["Hospital"].str.title()
df["Doctor"] = df["Doctor"].str.title()
df["Medical Condition"] = df["Medical Condition"].str.title()
df["Insurance Provider"] = df["Insurance Provider"].str.title()
```

```python
df.head()
```

```python
 

plt.figure(figsize=(6,4))
plt.boxplot(df["Age"])
plt.title("Boxplot of Age")
plt.ylabel("Age")
plt.show()
```

```python
df["Length_of_Stay"] = (df["Discharge Date"]-df["Date of Admission"]).dt.days
```

```python
df.head()
```

```python
df.shape
```

```python
bins=[0,18,35,50,65,100]

labels=["Child",
        "Young Adult",
        "Adult",
        "Senior",
        "Elder"]

df["Age_Group"]=pd.cut(df["Age"],
bins=bins,
labels=labels)
```

```python
df.head()
```

```python
bins = [0,13000,37000,53000]

labels=["Low",
        "medium",
        "High"]
df["Billing Category"] = pd.cut(
    df["Billing Amount"],
    bins = bins,
    labels = labels)
                                
```

```python
df.head()
```

```python
df.shape
```

```python
df.to_csv("healthcare_cleaned.csv",
index=False
)
```
